# Module 01 — Introduction to OpenEnv

Hands-on walkthrough of the OpenEnv lifecycle.

## Setup

```python
import sys, os
sys.path.insert(0, os.path.join(os.getcwd(), '../../src'))
print('Path set up.')
```

## Step 1: Import the Environment

```python
from openenv_env import CodeDebugEnvironment, Action
env = CodeDebugEnvironment(random_seed=42)
print('Environment created:', env)
```

## Step 2: Reset — start a new episode

```python
obs = env.reset(task_id="task_syntax_001")
print("task_id :", obs.task_id)
print("error_type:", obs.error_type)
print("difficulty:", obs.difficulty)
print("\nBuggy code:\n", obs.buggy_code)
```

## Step 3: Inspect the Observation fields

```python
print("step      :", obs.step)
print("max_steps :", obs.max_steps)
print("done      :", obs.done)
print("reward    :", obs.reward)
print("hints     :", obs.hints)
```

## Step 4: Submit an Action (wrong answer first)

```python
wrong_action = Action(fixed_code=obs.buggy_code, explanation="No change", confidence=0.1)
result = env.step(wrong_action)
print("reward:", result.reward)
print("done  :", result.done)
```

## Step 5: Submit the correct fix

```python
correct_action = Action(
    fixed_code="def multiply(a, b):\n    result = a * b\n    return result\n",
    explanation="Added the missing colon after the function signature.",
    confidence=0.95
)
result = env.step(correct_action)
print("reward:", result.reward)
print("done  :", result.done)
```

## Step 6: Check State

```python
state = env.state()
print("cumulative_reward:", state.cumulative_reward)
print("step             :", state.step)
print("history entries  :", len(state.history))
```
